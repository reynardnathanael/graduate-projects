```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameter Zoom (Interval < 0.001)
# Anda bisa mencoba koordinat lain untuk menemukan keindahan berbeda
center_x, center_y = -0.743643887037158, 0.131825904205311
zoom_width = 0.0005

x_start, x_end = center_x - zoom_width, center_x + zoom_width
y_start, y_end = center_y - zoom_width, center_y + zoom_width

# 2. Resolusi dan Iterasi (Tingkatkan nmax untuk detail lebih tajam)
N = 500
nmax = 200

x = np.linspace(x_start, x_end, N)
y = np.linspace(y_start, y_end, N)
H = np.zeros((N, N))

def f(C, max_iter):
    z = 0
    for i in range(max_iter):
        z = z**2 + C
        if np.abs(z) > 10: # limit bisa disesuaikan
            return i
    return max_iter

# 3. Hitung Matrix
for i in range(N):
    for j in range(N):
        # Perhatikan: j untuk sumbu y (imajiner), i untuk sumbu x (real)
        # Gunakan H[j, i] agar orientasi gambar benar (y di vertikal)
        H[j, i] = f(complex(x[i], y[j]), nmax)

# 4. Tampilkan dan Simpan
plt.figure(figsize=(10, 10), dpi=150)
plt.imshow(H, extent=[x_start, x_end, y_start, y_end], cmap='magma', interpolation='bilinear')
plt.colorbar(label='Iterations to divergence')
plt.title(f'Mandelbrot Zoom\nCenter: ({center_x}, {center_y}), nmax: {nmax}')
plt.xlabel('Re(C)')
plt.ylabel('Im(C)')

# Simpan gambar
plt.savefig('mandelbrot_beautiful_zoom.png')
plt.show()

print(f'Gambar telah disimpan sebagai mandelbrot_beautiful_zoom.png')
print(f'Parameter: x=[{x_start}, {x_end}], y=[{y_start}, {y_end}], nmax={nmax}')
```
