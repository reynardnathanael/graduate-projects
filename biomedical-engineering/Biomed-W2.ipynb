```python
n = 501 # diameter
r = n / 2.0 # jari-jari = 250.5
data = np.zeros((n, n)) # dimension: 500 x 500

# Loop through every pixel to check if it's inside the circle
for i in range(n):
  for j in range(n):
    # Calculate distance from center (r, r)
    if (i - r)**2 + (j - r)**2 < r**2: # inside circle or not? if < , yes
      data[i, j] = 1

plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
H4 = np.loadtxt('topography_180x360_grid.txt')

# # Set elevations above 2000m to 10000m
# H4[H4 > 2000] = 10000

# # Set depths below -2000m to -10000m
# H4[H4 < -2000] = -10000

for i in range(0, 180):
  for j in range(0, 360):
    if H4[i,j] > 2000: # elevation above 2000m -> become 10000m
      H4[i,j] = 10000
    elif H4[i,j] < -2000: # depths below -2000m -> become -10000m
      H4[i,j] = -10000

plt.imshow(H4, interpolation='nearest')
plt.show()
```

```python
H4 = np.loadtxt('topography_180x360_grid.txt')

# # Set elevations above 2000m to 10000m
# H4[H4 > 2000] = 10000

# # Set depths below -2000m to -10000m
# H4[H4 < -2000] = -10000

# # Use logical AND (&) with parentheses for range checks in NumPy
# H4[(H4 > -2000) & (H4 < 2000)] = 0

for i in range(0, 180):
  for j in range(0, 360):
    if H4[i,j] > 2000: # elevation above 2000m -> become 10000m
      H4[i,j] = 10000
    elif H4[i,j] < -2000: # depths below -2000m -> become -10000m
      H4[i,j] = -10000
    else: # between -2000m and +2000m -> become 0
      H4[i,j] = 0

plt.imshow(H4, interpolation='nearest')
plt.show()
```

```python
#Accomodating Wakanda
H5 = np.copy(H4);

for i in range(0, 180):
  for j in range(0, 360):
    if j > 150:
      # takes pixels from the left side (columns 0–209) and moves them 150 units to the right.
      H5[i,j] = H4[i,j-150]
    else:
      # This takes pixels from the right side and wraps them around to the beginning (columns 0-150).
      H5[i,j] = H4[i,j+209]

plt.imshow(H5, interpolation='nearest')
plt.show()
```
