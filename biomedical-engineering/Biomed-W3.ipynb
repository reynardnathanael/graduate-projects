Compute and print the variance and the standard deviation sigma between your data set and the solution as
described on the lecture notes.

```python
# variance = np.var(diff)
# standard_deviation = np.std(diff)
```

```python
# using Population Variance -> σ² = ∑(Xi - μ)²/N
avg = diff.sum()/nNumbers # μ

sx = 0
# range i=0 sampai i=99
for i in range(0, nNumbers): # ∑(Xi - μ)²
  sx += (diff[i] - avg)**2 # (Xi - μ)²

variance = sx/(nNumbers) # σ² = ∑(Xi - μ)²/N
standard_deviation = np.sqrt(variance) # σ = √(∑(Xi - μ)²/N)

print(f'Variance: {variance}')
print(f'Standard Deviation (sigma): {standard_deviation}')
```

```python
# using Sample Variance -> s² = ∑(xi - x̄)²/n-1
avg_2 = diff.sum()/nNumbers # x̄

sx_2 = 0
for j in range(0, nNumbers): # ∑(xi - x̄)²
  sx_2 += (diff[j] - avg_2)**2 # (xi - x̄)²

variance_2 = sx_2/(nNumbers-1) # s² = ∑(xi - x̄)²/n-1
standard_deviation_2 = np.sqrt(variance_2) # s = √(∑(xi - x̄)²/n-1)

print(f'Variance: {variance_2}')
print(f'Standard Deviation (sigma): {standard_deviation_2}')
```
